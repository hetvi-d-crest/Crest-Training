```python
from langchain_core.runnables import RunnableLambda
```

```python
find_sum = lambda x: sum(x)
```

```python
find_sum([1,2,5])
```

```python
find_square = lambda x : x**2
```

```python
find_square(8)
```

```python
runnable_sum = RunnableLambda(find_sum)
```

```python
runnable_sum.invoke([1,2,5])
```

```python
runnable_square = RunnableLambda(find_square)
```

```python
runnable_square.invoke(8)
```

```python
chain = runnable_sum | runnable_square
```

```python
chain.invoke([1,2,5])
```

```python
chain.get_graph().print_ascii()
```
