### Fake Data

```python
df = pd.DataFrame(np.random.randn(100,4),columns='A B C D'.split())
```

```python
df.head()
```

```python
df2 = pd.DataFrame({'Category':['A','B','C'],'Values':[32,43,50]})
```

```python
df2.head()
```

## Scatter

```python
px.scatter(df, x='A', y='B', size_max=10)
```

```python
df.plot(kind ='scatter', x= 'A', y='B')
```

## Bar Plots

```python
px.bar(df2, x = 'Category', y='Values')
```

```python
pd.options.plotting.backend = "plotly"
```

```python
df.count().plot(kind='bar')
```

## Boxplots

```python
df.plot(kind='box')
```

```python
px.box(df)
```

## 3d Surface

```python
df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})
```

```python
px.scatter_3d(df3, x='x', y='y', z='z', color='z', color_continuous_scale='RdYlBu')
```

## Spread

```python
df[['A','B']].plot(kind='spread')
```

## histogram

```python
px.histogram(df, x='A', nbins=25)
```

```python
df['C_abs'] = df['C'].abs()

fig = px.scatter(df, x='A', y='B', 
                 size='C_abs', 
                 color='C', # Use the original 'C' for color to show +/-
                 size_max=20,
                 title="Bubble Size by Magnitude of C")
fig.show()
```
