```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
connect_tokens = ['connect', 'connected', 'connectivity', 'connecting', 'connects']
```

```python
for t in connect_tokens:
    print(t, ": ", ps.stem(t))
```

```python
learn_tokens = ['learned', 'learning', 'learners', 'learn', 'learns', 'learner']
```

```python
for t in learn_tokens:
    print(t, ": ", ps.stem(t))
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

```python
for t in likes_tokens:
    print(t, ": ", ps.stem(t))
```
