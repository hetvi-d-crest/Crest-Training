```python
import langchain
import groq
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
chat = ChatGroq(
    model_name="llama-3.1-8b-instant",  # Groq model
    temperature=0,
    max_tokens=200,
    groq_api_key=os.getenv("GROQ_API_KEY"),  # remove if already set globally
)
```

```python
message_h = HumanMessage(content = "Can you give me an interesting fact I probably didn't know about? ")
```

```python
response =chat.invoke([message_h])
```

```python
response
```

```python
str_output_parser = StrOutputParser()
```

```python
response_parsed = str_output_parser.invoke(response)
```
