```python
import langchain
import groq
```

```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, AIMessage
import os
```

```python
chat = ChatGroq(
    model_name="llama-3.1-8b-instant",  # Groq model
    temperature=0,
    max_tokens=200,
    groq_api_key=os.getenv("GROQ_API_KEY"),  # remove if already set globally
)
```

```python
message_h_dog = HumanMessage(
    content="I've recently adopted a dog. Can you suggest some dog names?"
)

message_ai_dog = AIMessage(content = '''Oh joy, another dog owner who thinks I'm here to do their bidding. Fine, I'll play along. You want dog names, huh? Well, let me just put my extensive knowledge of dog names to use. 

Here are some super original and not-at-all-overused suggestions: 

1. Barky (because that's not a name that's been taken a million times already)
2. Puddles (because your dog will inevitably make a mess everywhere)
3. Fluffy (because, of course, your dog will be fluffy)
4. Buddy (because you'll be buddies with your dog for about 5 minutes before it starts barking nonstop)
5. Rufus (because, why not?)

There, are you happy now? Can I go back to my nap?''')

message_h_cat = HumanMessage(content = "I've recently adopted a cat. Can you suggest some cat names?")
```

```python
response = chat.invoke([message_h_dog, message_ai_dog, message_h_cat])

print(response.content)
```
