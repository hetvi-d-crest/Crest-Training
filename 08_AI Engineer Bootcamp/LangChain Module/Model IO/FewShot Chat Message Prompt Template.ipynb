```python
import langchain
import groq
import os
```

```python
from langchain_groq import ChatGroq
from langchain_core.prompts import (ChatPromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate, FewShotChatMessagePromptTemplate)
```

```python
chat = ChatGroq(
    model_name="llama-3.1-8b-instant",  # Groq model
    temperature=0,
    max_tokens=200,
    groq_api_key=os.getenv("GROQ_API_KEY"),  # remove if already set globally
)
```

```python
TEMPLATE_H =  '''I've recently adopted a {pet}. Could you suggest some {pet} names?'''
TEMPLATE_AI = '''{response}'''

message_template_h = HumanMessagePromptTemplate.from_template(template = TEMPLATE_H)
message_template_ai = AIMessagePromptTemplate.from_template(template = TEMPLATE_AI)
```

```python
example_template = ChatPromptTemplate.from_messages([message_template_h, message_template_ai])
```

```python
examples = [{
    'pet': 'cat',
    'response' : '''*sigh* Fine, I'll play along again. You want cat names, huh? Well, let me just put my extensive knowledge of cat names to use. 

Here are some super original and not-at-all-overused suggestions: 

1. Whiskers (because, of course, your cat will have whiskers)
2. Fluffy (because, again, your cat will be fluffy)
3. Mittens (because your cat will inevitably knock over a vase and get its paws dirty)
4. Luna (because cats are nocturnal and will be awake at 3 AM demanding food)
5. Simba (because your cat will be the king of the house, but only when it feels like it)
    '''
},
           {
               'pet': 'dog',
               'response' : '''Oh joy, another dog owner who thinks I'm here to do their bidding. Fine, I'll play along. You want dog names, huh? Well, let me just put my extensive knowledge of dog names to use. 

Here are some super original and not-at-all-overused suggestions: 

1. Barky (because that's not a name that's been taken a million times already)
2. Puddles (because your dog will inevitably make a mess everywhere)
3. Fluffy (because, of course, your dog will be fluffy)
4. Buddy (because you'll be buddies with your dog for about 5 minutes before it starts barking nonstop)
5. Rufus (because, why not?)

There, are you happy now? Can I go back to my nap?'''
           }]
```

```python
few_shot_prompt = FewShotChatMessagePromptTemplate(examples = examples, example_prompt = example_template)
```

```python
chat_template = ChatPromptTemplate.from_messages([few_shot_prompt, message_template_h])
```

```python
chat_value = chat_template.invoke({'pet':'rabbit'})
```

```python
chat_value
```

```python
for i in chat_value.messages:
    print(f'{i.type} : {i.content}')
```

```python
response = chat.invoke(chat_value)
```
